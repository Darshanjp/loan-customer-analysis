```python
#1.top 5 loan amounts
loan_details=pd.merge(cust_data,loan_data,on="customer_id", how ="inner")
loan_details
```

```python
#top 5 loans amount --pd.grouphy()
loan_details.groupby("customer_name")["loan_amount"].sum().sort_values(ascending=False).head(5)
```

```python
##2.lowest
loan_details.groupby("customer_name")["loan_amount"].sum().sort_values(ascending=False).tail(5)
```

```python
#3 cities with most number of loan customers
loan_details.groupby("city")["customer_name"].count().sort_values(ascending=False)
```

```python
#4 approved,not approved,under review
loan_details.groupby("approval_status")["customer_name"].count().sort_values(ascending=False)
```

```python
# top 5 loan intersest rate
loan_details.groupby("customer_name")["interest_rate"].sum().sort_values(ascending=False).head(5)
```
